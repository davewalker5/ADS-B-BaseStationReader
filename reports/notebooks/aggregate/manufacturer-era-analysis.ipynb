# Manufacturer and Era Analysis

How are observed aircraft and their ages distributed across manufacturers and manufacturing eras?

```python
%run ../pathutils.ipynb
%run ../database.ipynb
%run ../export.ipynb
%run ../report-header.ipynb
%run ./aircraft-population-utils.ipynb

export_outputs = True
export_folder = get_export_folder_path()
```

```python
report_metadata = display_report_header('Manufacturer and Era Analysis')
population = load_aircraft_population()
aircraft = unique_aircraft(population)
manufacturer_summary = grouped_age_summary(population, 'Manufacturer')
era_distribution = pd.crosstab(aircraft['Manufacturer'], aircraft['Manufacturing Era'], dropna=False).reindex(columns=ERA_ORDER, fill_value=0)
manufacturer_summary.head(25)
```

```python
# Keep the chart legible by displaying the manufacturers with the most unique aircraft.
top_manufacturers = aircraft['Manufacturer'].value_counts().head(15).index
ax = era_distribution.loc[era_distribution.index.intersection(top_manufacturers)].plot.bar(stacked=True, figsize=(14, 7), title='Unique aircraft by manufacturer and era')
ax.set_ylabel('Unique aircraft')
ax.legend(title='Manufacturing era', bbox_to_anchor=(1.02, 1), loc='upper left')
if export_outputs:
    export_chart(export_folder, 'manufacturer-era-analysis', 'png')
    export_to_spreadsheet(export_folder, 'manufacturer-era-analysis.xlsx', {'Manufacturers': manufacturer_summary, 'Eras': era_distribution.reset_index(), 'Aircraft': aircraft})
```
