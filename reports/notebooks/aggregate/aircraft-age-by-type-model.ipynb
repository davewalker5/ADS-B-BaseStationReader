# Aircraft Age by Type and Model

Which aircraft types, models, and manufacturers contain the oldest observed aircraft?

```python
%run ../pathutils.ipynb
%run ../database.ipynb
%run ../export.ipynb
%run ../report-header.ipynb
%run ./aircraft-population-utils.ipynb

export_outputs = True
export_folder = get_export_folder_path()
```

```python
report_metadata = display_report_header('Aircraft Age by Type and Model')
population = load_aircraft_population()
type_summary = grouped_age_summary(population, 'ICAO Type')
model_summary = grouped_age_summary(population, 'Model')
manufacturer_summary = grouped_age_summary(population, 'Manufacturer')
type_summary.head(25)
```

```python
# Plot only the most represented types, using one known-age record per unique aircraft.
plot_data = unique_aircraft(population).dropna(subset=['Age At Observation'])
type_sample_sizes = plot_data['ICAO Type'].value_counts()
top_types = type_sample_sizes.head(12).index
ax = plot_data[plot_data['ICAO Type'].isin(top_types)].boxplot(column='Age At Observation', by='ICAO Type', figsize=(13, 7), rot=45)
# Put the plotted sample size directly on every category label to make narrow boxes safer to interpret.
ax.set_xticklabels([f'{label.get_text()} (n={type_sample_sizes[label.get_text()]})' for label in ax.get_xticklabels()])
ax.set_title('Age distributions for the most represented ICAO types')
ax.set_xlabel('ICAO type (n = unique aircraft with known age)')
ax.set_ylabel('Age (years)')
ax.figure.suptitle('')
if export_outputs:
    export_chart(export_folder, 'aircraft-age-by-type-model', 'png')
    export_to_spreadsheet(export_folder, 'aircraft-age-by-type-model.xlsx', {'ICAO Types': type_summary, 'Models': model_summary, 'Manufacturers': manufacturer_summary})
```
