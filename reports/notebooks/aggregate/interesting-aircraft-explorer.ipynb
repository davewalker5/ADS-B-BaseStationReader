# Interesting Aircraft Explorer

A discovery tool that surfaces candidates without automatically classifying them as important. Adjust the exploratory controls below to investigate different questions.

```python
%run ../pathutils.ipynb
%run ../database.ipynb
%run ../export.ipynb
%run ../report-header.ipynb
%run ./aircraft-population-utils.ipynb

export_outputs = True
export_folder = get_export_folder_path()
minimum_age = 60
maximum_model_aircraft = 1
single_session_only = False
```

```python
report_metadata = display_report_header('Interesting Aircraft Explorer')
population = load_aircraft_population()
candidates = rarity_candidates(population)
# Controls are deliberately transparent and editable rather than permanent labels.
selected = candidates[(candidates['Age At Observation'].fillna(-1) >= minimum_age) |
                      ((candidates['Model'] != 'Unknown') & (candidates['Model Aircraft Count'] <= maximum_model_aircraft))]
if single_session_only:
    selected = selected[selected['Sessions'] == 1]
selected
```

```python
# Candidate reasons are evidence prompts, not a judgement of significance.
if export_outputs:
    export_to_spreadsheet(export_folder, 'interesting-aircraft-explorer.xlsx', {'Candidates': selected, 'All Aircraft Signals': candidates})
selected[['Address', 'Registration', 'Model', 'Manufacturer', 'Manufacture Year', 'Age At Observation', 'Observations', 'Sessions', 'Candidate Reasons']].head(100)
```
