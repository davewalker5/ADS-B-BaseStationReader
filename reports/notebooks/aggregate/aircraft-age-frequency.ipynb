# Aircraft Age Versus Observation Frequency

Are older aircraft rare because they are uncommon, or because they are observed less frequently?

```python
%run ../pathutils.ipynb
%run ../database.ipynb
%run ../export.ipynb
%run ../report-header.ipynb
%run ./aircraft-population-utils.ipynb

export_outputs = True
export_folder = get_export_folder_path()
```

```python
report_metadata = display_report_header('Aircraft Age Versus Observation Frequency')
population = load_aircraft_population()
frequency = aircraft_frequency(population)
known = frequency.dropna(subset=['Age At Observation']).copy()
frequent_historic = known[known['Age At Observation'] >= 60].sort_values(['Sessions', 'Observations'], ascending=False)
rare_modern = known[(known['Age At Observation'] < 60) & (known['Sessions'] == 1)].sort_values('Observations')
frequent_historic.head(25)
```

```python
# Log scaling preserves visibility for aircraft with very different observation counts.
ax = known.plot.scatter(x='Age At Observation', y='Observations', c='Sessions', cmap='viridis', logy=True, figsize=(11, 7), title='Aircraft age versus observation frequency')
ax.axvline(60, color='firebrick', linestyle='--', label='Exploratory 60-year threshold')
ax.set_ylabel('Observation records (log scale)')
ax.legend()
if export_outputs:
    export_chart(export_folder, 'aircraft-age-frequency', 'png')
    export_to_spreadsheet(export_folder, 'aircraft-age-frequency.xlsx', {'Aircraft Frequency': frequency, 'Frequent Historic': frequent_historic, 'Rare Modern': rare_modern})
```
